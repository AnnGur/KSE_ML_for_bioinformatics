## 3. PCA Step 1 — Center the data


````python
X = expression_pca.to_numpy(dtype=float)
feature_means = X.mean(axis=0)
X_centered = X - feature_means


print(float(np.abs(X_centered.mean(axis=0)).max()))
````

## 4. Optional scaling

For each feature, unit-variance scaling is

\[
z_{ij} = \frac{x_{ij} - \mu_j}{\sigma_j}.
\]

Should every gene contribute equal variance? For this small expression example, we use log-CPM and select variable genes, but do not scale each gene to unit variance in the main PCA. Unit-variance scaling would give low-abundance/noisy genes the same influence as highly variable genes.


````python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("Centered-only range:", (X_centered.min(), X_centered.max()))
print("Scaled feature means (first 5):", X_scaled.mean(axis=0)[:5])
````

## 5. PCA Step 2 — Covariance

After centering, the sample covariance matrix is

\[
C = \frac{1}{n-1} X_c^{\mathsf{T}} X_c.
\]

````python
covariance_matrix = (X_centered.T @ X_centered) / (X_centered.shape[0] - 1)
covariance_matrix.shape
````

````python
covariance_matrix
````

## 6. PCA Step 3 — Eigenvalues and eigenvectors

The principal directions satisfy the eigenvalue equation

\[
C v_k = \lambda_k v_k.
\]

- \(v_k\): principal direction / loading vector
- \(\lambda_k\): variance captured along that direction

Because this dataset has only five samples, the centered matrix has at most four non-zero directions. Rather than diagonalizing the full 1,000 × 1,000 covariance matrix, we use the mathematically equivalent 5 × 5 sample-space covariance matrix and reconstruct the feature-space loading vectors.

````python
n_samples, n_features = X_centered.shape
sample_covariance = (X_centered @ X_centered.T) / (n_samples - 1)
sample_eigenvalues, sample_eigenvectors = np.linalg.eigh(sample_covariance)
order = np.argsort(sample_eigenvalues)[::-1]
sample_eigenvalues = sample_eigenvalues[order]
sample_eigenvectors = sample_eigenvectors[:, order]
sample_eigenvectors
````

````python

# Recover feature-space eigenvectors from the small eigendecomposition.
positive = sample_eigenvalues > 1e-10
eigenvalues = sample_eigenvalues[positive]
eigenvectors = (X_centered.T @ sample_eigenvectors[:, positive]) / np.sqrt((n_samples - 1) * eigenvalues)

residual = np.linalg.norm(covariance_matrix @ eigenvectors[:, 0] - eigenvalues[0] * eigenvectors[:, 0])
print("Diagonalized matrix shape:", sample_covariance.shape)
print("Non-zero components:", len(eigenvalues))
print("Largest eigenvalues:", eigenvalues)
print("First eigenvector equation residual:", residual)
````

## 7. Explained variance

````python
explained_variance_ratio = eigenvalues / eigenvalues.sum()
cumulative_variance = np.cumsum(explained_variance_ratio)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(np.arange(1, len(eigenvalues) + 1), explained_variance_ratio, marker="o")
axes[0].set(xlabel="Principal component", ylabel="Explained variance ratio", title="Scree plot")
axes[1].plot(np.arange(1, len(eigenvalues) + 1), cumulative_variance, marker="o")
axes[1].axhline(0.90, color="grey", linestyle="--", label="90%")
axes[1].set(xlabel="Number of components", ylabel="Cumulative explained variance", title="Cumulative variance")
axes[1].set_ylim(0, 1.05)
axes[1].legend()
plt.tight_layout()
print("PC1:", round(float(explained_variance_ratio[0]), 3))
print("PC1 + PC2:", round(float(cumulative_variance[1]), 3))
````

## 8. PCA Step 4 — Projection

\[
Z = X_cV
\]

Rows of \(Z\) are sample **scores**.


````python
scores = X_centered @ eigenvectors
pca_manual = pd.DataFrame(scores[:, :2], index=expression_pca.index, columns=["PC1", "PC2"])
display(pca_manual)
````

# Part III — KNN and SVM

## 18. Supervised dataset

**Dataset:** Wisconsin Diagnostic Breast Cance

This dataset contains measurements computed from digitized images of fine-needle aspirate (FNA) samples of breast masses. The prediction target is a diagnosis label. We use it to study distance-based and margin-based classifiers; the goal is methodological, not to build a clinical decision system.

- Samples: 569 nuclei-derived observations
- Features: 30 numeric measurements derived from 10 cell-nucleus descriptors
- Target: malignant or benign diagnosis
- Class balance: 212 malignant and 357 benign observations
- Biological meaning: geometric and texture properties of cell nuclei from an FNA image

KNN classifies a point using nearby training observations. With Euclidean distance,

\[
d(x_i, x_j) = \sqrt{\sum_{r=1}^{p}(x_{ir}-x_{jr})^2}.
\]

An SVM instead learns a decision boundary with a large margin. For a soft-margin linear SVM, the parameter \(C\) controls the trade-off between a wide margin and training violations:

\[
\min_{w,b,\xi} \frac{1}{2}\lVert w\rVert^2 + C\sum_i \xi_i.
\]

We will keep the test set untouched until the final comparison.

## 21. KNN without scaling

````python
def classifier_metrics(model, X_eval, y_eval):
    predictions = model.predict(X_eval)
    return {
        "accuracy": accuracy_score(y_eval, predictions),
        "balanced_accuracy": balanced_accuracy_score(y_eval, predictions),
        "macro_f1": f1_score(y_eval, predictions, average="macro")}

knn_raw = KNeighborsClassifier(n_neighbors=5)
knn_raw.fit(X_train, y_train)
raw_knn_metrics = classifier_metrics(knn_raw, X_valid, y_valid)
display(pd.Series(raw_knn_metrics, name="raw KNN").to_frame())
````

## 22. KNN with scaling

````python
knn = Pipeline([
    ("scale", StandardScaler()),
    ("model", KNeighborsClassifier(n_neighbors=5))])
knn.fit(X_train, y_train)
scaled_knn_metrics = classifier_metrics(knn, X_valid, y_valid)
display(pd.DataFrame([raw_knn_metrics, scaled_knn_metrics], index=["raw KNN", "scaled KNN"]))
````

## 23. Manual selection of k

````python
k_values = [1, 3, 5, 7, 11, 21]

knn_results = []
for k in k_values:
    candidate = Pipeline([
        ("scale", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=k))])
    candidate.fit(X_train, y_train)
    metrics = classifier_metrics(candidate, X_valid, y_valid)
    knn_results.append({"k": k, **metrics})

knn_results = pd.DataFrame(knn_results)
display(knn_results)
best_k = int(knn_results.loc[knn_results["macro_f1"].idxmax(), "k"])

plt.figure(figsize=(7, 4))
for metric in ["accuracy", "balanced_accuracy", "macro_f1"]:
    plt.plot(knn_results["k"], knn_results[metric], marker="o", label=metric)
plt.axvline(best_k, color="black", linestyle="--", alpha=0.5, label=f"selected k={best_k}")
plt.xlabel("Number of neighbours, k")
plt.ylabel("Validation score")
plt.title("Manual KNN selection")
plt.legend()
plt.tight_layout()
plt.show()
````

## 24. Linear SVM

A linear SVM searches for a separating hyperplane. The parameter **C** controls the penalty for observations that are inside the margin or misclassified:

- small **C**: wider margin, more tolerance for training errors;
- large **C**: narrower margin, stronger pressure to fit the training data.

Because the objective uses distances and dot products, scaling belongs inside the pipeline.

````python
svm_linear = Pipeline([
    ("scale", StandardScaler()),
    ("model", SVC(kernel="linear", C=0.01)),
])
svm_linear.fit(X_train, y_train)
linear_svm_metrics = classifier_metrics(svm_linear, X_valid, y_valid)
display(pd.Series(linear_svm_metrics, name="linear SVM").to_frame())
````

## 25. RBF and polynomial SVM kernels

The radial basis function (RBF) kernel compares observations through their distance:

\[
K_{\mathrm{RBF}}(x_i, x_j) = \exp\left(-\gamma \lVert x_i - x_j \rVert^2\right).
\]

Here **gamma** controls how local the influence of one observation is:

- small **gamma**: smoother, broader influence;
- large **gamma**: very local influence and more flexible boundaries.

The polynomial kernel is

\[
K_{\mathrm{poly}}(x_i, x_j) = \left(\gamma x_i^{\mathsf{T}}x_j + r\right)^d,
\]

where \(d\) is the polynomial degree and \(r\) is the intercept term (`coef0` in scikit-learn).

The next illustration uses two-dimensional representations only to make decision boundaries visible. The final comparison trains each model on the complete feature set.

````python
svm_rbf = Pipeline([
    ("scale", StandardScaler()),
    ("model", SVC(kernel="rbf", C=1, gamma=0.01)),
])
svm_rbf.fit(X_train, y_train)
rbf_svm_metrics = classifier_metrics(svm_rbf, X_valid, y_valid)
display(pd.Series(rbf_svm_metrics, name="RBF SVM").to_frame())
````

````python
poly_rbf = Pipeline([
    ("scale", StandardScaler()),
    ("model", SVC(kernel="poly", C=10, degree=3)),
])
poly_rbf.fit(X_train, y_train)
poly_svm_metrics = classifier_metrics(poly_rbf, X_valid, y_valid)
display(pd.Series(poly_svm_metrics, name="Poly SVM").to_frame())
````

## 26. Visual illustration of the SVM kernels

To draw a decision boundary, we temporarily represent the training observations with two coordinates. PCA is a linear representation. The classifiers are trained on the displayed two-dimensional coordinates, so these plots are not the final performance evaluation.

We compare three kernels: linear, polynomial, and RBF.

````python
X_train_numeric = X_train.to_numpy(dtype=float)
y_train_numeric = y_train.to_numpy()
visual_scaler = StandardScaler()
X_train_scaled_for_visualization = visual_scaler.fit_transform(X_train_numeric)

visual_pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_train_pca2 = visual_pca.fit_transform(X_train_scaled_for_visualization)
visual_tsne = TSNE(n_components=2, perplexity=30, init="pca", learning_rate="auto", max_iter=750, random_state=RANDOM_STATE)
X_train_tsne2 = visual_tsne.fit_transform(X_train_scaled_for_visualization)

def plot_svm_boundary(ax, X_2d, y_2d, kernel, title):
    parameters = {"kernel": kernel, "C": 1, "gamma": "scale"}
    if kernel == "poly":
        parameters.update({"degree": 3, "coef0": 1})
    model = SVC(**parameters)
    model.fit(X_2d, y_2d)
    x_min, x_max = X_2d[:, 0].min() - 1, X_2d[:, 0].max() + 1
    y_min, y_max = X_2d[:, 1].min() - 1, X_2d[:, 1].max() + 1
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 250), np.linspace(y_min, y_max, 250))
    grid = np.c_[xx.ravel(), yy.ravel()]
    decision = model.decision_function(grid).reshape(xx.shape)
    ax.contourf(xx, yy, decision > 0, alpha=0.18, cmap="coolwarm")
    ax.contour(xx, yy, decision, levels=[0], colors="black", linewidths=1.4)
    ax.scatter(X_2d[:, 0], X_2d[:, 1], c=y_2d, cmap="coolwarm", s=16, edgecolor="white", linewidth=0.25)
    ax.set_title(title)
    ax.set_xticks([])
    ax.set_yticks([])

kernels = ["linear", "poly", "rbf"]
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for column, kernel in enumerate(kernels):
    plot_svm_boundary(axes[0, column], X_train_pca2, y_train_numeric, kernel, f"{kernel} kernel on PCA(2)")
    plot_svm_boundary(axes[1, column], X_train_tsne2, y_train_numeric, kernel, f"{kernel} kernel on t-SNE(2)")
axes[0, 0].set_ylabel("PCA(2)")
axes[1, 0].set_ylabel("t-SNE(2)")
plt.tight_layout()
````

## 34. Choosing K

The elbow plot shows within-cluster sum of squares (inertia). Silhouette score compares within-cluster cohesion with separation from the nearest alternative cluster. Neither is a substitute for biological interpretation.

````python
k_values = range(2, 9)
inertias = []
silhouettes = []
for k in k_values:
    candidate = KMeans(n_clusters=k, n_init=10, random_state=RANDOM_STATE)
    labels = candidate.fit_predict(X_micro_pca)
    inertias.append(candidate.inertia_)
    silhouettes.append(silhouette_score(X_micro_pca, labels, sample_size=min(5000, len(X_micro_pca)), random_state=RANDOM_STATE))

cluster_diagnostics = pd.DataFrame({"K": list(k_values), "inertia": inertias, "silhouette": silhouettes})
display(cluster_diagnostics)
selected_k = int(cluster_diagnostics.loc[cluster_diagnostics["silhouette"].idxmax(), "K"])
print("Selected K from the largest silhouette score:", selected_k)
````

## 38.2 PCA diagnosis: batch or biology?

For each of the first PCs we compute the share of variance explained by `source` and by `label` (η², i.e. between-group sum of squares / total sum of squares). A PC with high η² for `source` and low η² for `label` is a technical axis.

````python
def eta_squared(values, groups):
    values = np.asarray(values, dtype=float)
    total = ((values - values.mean()) ** 2).sum()
    between = sum(len(values[groups == g]) * (values[groups == g].mean() - values.mean()) ** 2
                  for g in np.unique(groups))
    return between / total

eta = pd.DataFrame({
    "PC": [f"PC{i+1}" for i in range(10)],
    "variance explained": pca_ch.explained_variance_ratio_[:10],
    "eta2 source": [eta_squared(Z_ch[:, i], ch_source) for i in range(10)],
    "eta2 label": [eta_squared(Z_ch[:, i], ch_label) for i in range(10)],
}).set_index("PC")
display(eta.round(3))

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, (a, b) in zip(axes, [(0, 1), (1, 2), (2, 3)]):
    for s in np.unique(ch_source):
        m = ch_source == s
        ax.scatter(Z_ch[m, a], Z_ch[m, b], s=4, alpha=0.4, label=s)
    ax.set(xlabel=f"PC{a+1}", ylabel=f"PC{b+1}", title="PCA colored by source")
axes[0].legend(markerscale=3)
plt.tight_layout()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, (a, b) in zip(axes, [(0, 1), (1, 2), (2, 3)]):
    for l in np.unique(ch_label):
        m = ch_label == l
        ax.scatter(Z_ch[m, a], Z_ch[m, b], s=4, alpha=0.4, label=l)
    ax.set(xlabel=f"PC{a+1}", ylabel=f"PC{b+1}", title="PCA colored by label")
axes[0].legend(markerscale=3)
plt.tight_layout()
````

### Interpretation (PCA)
- PC1 explains the most variance (16%) but is neither source nor label (η² 0.02 and 0.004), so it is probably general image appearance (cell size, intensity, staining).
- PC2 (η² source = 0.45) and PC3 (0.26) are batch axes: nearly half of PC2's variance is explained by which lab the cell came from. PC4 (0.18), PC8 and PC9 (about 0.13) also carry some source signal.
- The class label never exceeds η² = 0.09 on any PC (PC2 0.085 is the max), so the biological signal is much weaker than the technical one in the linear structure of the data.
- The PCA plots agree: colouring by `source` shows distinct clouds (especially `jimt` against the smaller sources), whereas `label` colours are mixed.
- Verdict: batch effect is visible

## 38.3 PCA vs t-SNE vs UMAP

`jimt` (22k cells) would dominate any projection and make t-SNE slow, so we draw a subset **stratified by source × label** (at most 400 cells per group), keeping the rare groups intact.

````python
rng_ch = np.random.default_rng(RANDOM_STATE)
group_key = pd.Series(ch_source) + "_" + pd.Series(ch_label)
sub_idx = np.concatenate([
    rng_ch.choice(idx, size=min(400, len(idx)), replace=False)
    for _, idx in group_key.groupby(group_key).indices.items()])
Z_sub, src_sub, lab_sub = Z_ch[sub_idx], ch_source[sub_idx], ch_label[sub_idx]
print("Subset size:", len(sub_idx))

tsne_ch = TSNE(n_components=2, perplexity=30, init="pca", learning_rate="auto",
               max_iter=750, random_state=RANDOM_STATE).fit_transform(Z_sub)
umap_ch = umap.UMAP(n_neighbors=15, min_dist=0.1, random_state=RANDOM_STATE, n_jobs=1).fit_transform(Z_sub)

projections = {"PCA": Z_sub[:, :2], "t-SNE": tsne_ch, "UMAP": umap_ch}
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for col, (name, emb) in enumerate(projections.items()):
    for row, (values, title) in enumerate([(lab_sub, "label"), (src_sub, "source")]):
        for v in np.unique(values):
            m = values == v
            axes[row, col].scatter(emb[m, 0], emb[m, 1], s=8, alpha=0.5, label=v)
        axes[row, col].set_title(f"{name} - {title}")
        axes[row, col].set_xticks([]); axes[row, col].set_yticks([])
        axes[row, col].legend(markerscale=2)
plt.tight_layout()
````

## 38.4 Validate the projections against the labels (quantitatively)

Visual separation can mislead. We score each 2D projection with the **silhouette** of `label` and of `source`, and with the fraction of each point's 15 nearest neighbours (in the projection) that share its label / source. If a projection separates `source` better than `label`, the structure it shows is mostly technical.

````python
from sklearn.neighbors import NearestNeighbors

def neighbour_agreement(emb, groups, k=15):
    idx = NearestNeighbors(n_neighbors=k + 1).fit(emb).kneighbors(emb, return_distance=False)[:, 1:]
    return float((groups[idx] == groups[:, None]).mean())

rows = []
for name, emb in projections.items():
    rows.append({
        "projection": name,
        "silhouette label": silhouette_score(emb, lab_sub),
        "silhouette source": silhouette_score(emb, src_sub),
        "kNN agreement label": neighbour_agreement(emb, lab_sub),
        "kNN agreement source": neighbour_agreement(emb, src_sub),
    })
display(pd.DataFrame(rows).set_index("projection").round(3))
````

### Interpretation (projections)
All three projections have a negative value for label (-0.06 to -0.09), so classes are not separated. 
For source, the profile is slightly negative in PCA (-0.03) but positive in t-SNE (0.34) and UMAP (0.37)\
The 15-nearest-neighbour agreement - same: 0.69-0.70 for label (t-SNE, UMAP) vs 0.95-0.96 for source.
The nonlinear methods make the batch structure look like clusters, while the label stays mixed.

## 38.5 KNN and SVM: predicting `label`

Two validation schemes:

1. **Random stratified split** - what we usually do. Cells from the same source appear in train and test.
2. **Leave-one-source-out** - train on all sources but one, test on the held-out source. This asks whether the model learned biology that transfers to a new lab.

A large gap between the two means the model is exploiting source-specific signal. Because classes are imbalanced we report balanced accuracy and macro-F1. Source `jimt` has no `S` cells and `chroma` has 13 cells, so held-out evaluation is done on the sources with enough data of each class (`alamar`, `mtt`, `study18`, plus `jimt` for C vs I); per-source scores are reported separately. To keep runtime reasonable the training set is capped at 1,500 cells per source x label group.

````python
from sklearn.model_selection import StratifiedKFold, cross_val_predict

# cap jimt dominance for training speed / balance
rng_m = np.random.default_rng(RANDOM_STATE)
keep = np.concatenate([
    rng_m.choice(idx, size=min(1500, len(idx)), replace=False)
    for _, idx in group_key.groupby(group_key).indices.items()])
X_m = challenge[ch_vgg].to_numpy(dtype=np.float32)[keep]
y_m, s_m = ch_label[keep], ch_source[keep]
print("Modelling subset:", X_m.shape, dict(pd.Series(y_m).value_counts()))

model_zoo = {
    "KNN (k=15)": Pipeline([("scale", StandardScaler()), ("pca", PCA(n_components=30, random_state=RANDOM_STATE)),
                            ("model", KNeighborsClassifier(n_neighbors=15))]),
    "SVM linear": Pipeline([("scale", StandardScaler()), ("pca", PCA(n_components=30, random_state=RANDOM_STATE)),
                            ("model", SVC(kernel="linear", C=1, class_weight="balanced"))]),
    "SVM RBF": Pipeline([("scale", StandardScaler()), ("pca", PCA(n_components=30, random_state=RANDOM_STATE)),
                         ("model", SVC(kernel="rbf", C=10, gamma="scale", class_weight="balanced"))]),
}

random_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
random_rows = []
oof_random = {}
for name, model in model_zoo.items():
    pred = cross_val_predict(model, X_m, y_m, cv=random_cv, n_jobs=-1)
    oof_random[name] = pred
    random_rows.append({"model": name,
                        "accuracy": accuracy_score(y_m, pred),
                        "balanced_accuracy": balanced_accuracy_score(y_m, pred),
                        "macro_f1": f1_score(y_m, pred, average="macro")})
print("Random stratified 5-fold CV (PCA fitted inside each fold):")
display(pd.DataFrame(random_rows).set_index("model").round(3))
````

````python
# Leave-one-source-out
loso_rows = []
for held_out in ["alamar", "mtt", "study18", "jimt"]:
    test_mask = s_m == held_out
    for name, model in model_zoo.items():
        fitted = clone(model).fit(X_m[~test_mask], y_m[~test_mask])
        pred = fitted.predict(X_m[test_mask])
        loso_rows.append({"held-out source": held_out, "model": name, "n": int(test_mask.sum()),
                          "accuracy": accuracy_score(y_m[test_mask], pred),
                          "balanced_accuracy": balanced_accuracy_score(y_m[test_mask], pred),
                          "macro_f1": f1_score(y_m[test_mask], pred, average="macro")})
loso = pd.DataFrame(loso_rows)
display(loso.round(3))
display(loso.pivot(index="held-out source", columns="model", values="balanced_accuracy").round(3))

best_name = pd.DataFrame(random_rows).sort_values("macro_f1").iloc[-1]["model"]
ConfusionMatrixDisplay.from_predictions(y_m, oof_random[best_name])
plt.title(f"Out-of-fold confusion matrix - {best_name}")
plt.show()
print(classification_report(y_m, oof_random[best_name]))
````

### Interpretation (supervised)
- Random 5-fold CV looks good: RBF SVM reaches balanced accuracy 0.79 and macro-F1 0.72; KNN reaches 0.64 and the linear SVM 0.73. The RBF SVM is the best model, so the class signal is nonlinear. The linear SVM has poor accuracy (0.64) because `class_weight="balanced"` trades accuracy for recall of the rare class `S`.
- The weak class is `S` (276 cells): recall 0.76 but precision only 0.37, so many `C`/`I` cells are labelled `S`. `C` and `I` are separated well (F1 0.81 and 0.85).
- Leave-one-source-out collapses: balanced accuracy is 0.36-0.54 for every held-out source and model, close to the 0.33-0.5 expected from guessing (for `jimt`, which only contains C and I, chance is 0.5, and the models score 0.44-0.48, i.e. at or below chance).
- So most of the random-CV performance comes from the model recognising source-specific appearance (and the correlation of source with class), not from a class signal that transfers to a new lab. Random CV is optimistic here and should not be reported as expected performance on new data.
- Caveats: `S` exists only in `alamar`, `mtt`, `study18` and `chroma`, so for these held-out sources the training set loses a large share of the `S` cells; and the training set was capped at 1,500 cells per source x label group.

## 38.6 Clustering with two distinct methods

We cluster in the 30-PC space (K-means and Ward agglomerative clustering). Clusters are compared with `label` **and** `source` using the adjusted Rand index (ARI) and normalized mutual information (NMI). The comparison with `source` is a batch check: a clustering that matches `source` better than `label` has found technical groups.

````python
Z_cl = Z_ch[keep][:, :20]
n_cl = 3  # three biological classes C / I / S

km_labels = KMeans(n_clusters=n_cl, n_init=10, random_state=RANDOM_STATE).fit_predict(Z_cl)
agg_labels = AgglomerativeClustering(n_clusters=n_cl, linkage="ward").fit_predict(Z_cl)

def compare(name, labels, idx):
    return {"method": name,
            "silhouette": silhouette_score(Z_cl[idx], labels, sample_size=min(5000, len(idx)), random_state=RANDOM_STATE),
            "ARI label": adjusted_rand_score(y_m[idx], labels),
            "NMI label": normalized_mutual_info_score(y_m[idx], labels),
            "ARI source": adjusted_rand_score(s_m[idx], labels),
            "NMI source": normalized_mutual_info_score(s_m[idx], labels)}

cluster_scores = pd.DataFrame([
    compare("K-means (K=3)", km_labels, np.arange(len(km_labels))),
    compare("Agglomerative Ward (K=3)", agg_labels, np.arange(len(agg_labels))),
]).set_index("method")
display(cluster_scores.round(3))

print("K-means clusters vs label:")
display(pd.crosstab(km_labels, y_m, rownames=["cluster"]))
print("K-means clusters vs source:")
display(pd.crosstab(km_labels, s_m, rownames=["cluster"]))
````

### Interpretation (clustering)
- K-means (K=3) and Ward agglomerative clustering both agree poorly with `label` (ARI 0.04 and 0.05, NMI 0.03 and 0.08) but substantially with `source` (ARI 0.54 and 0.81, NMI 0.44 and 0.74).
- The K-means crosstab shows it: cluster 0 contains all 3,000 `jimt` cells, cluster 1 is mainly `study18` plus `chroma`, and cluster 2 mixes `alamar`, `mtt` and `study18`. Each cluster contains all three labels in ~ dataset proportions.
- Two different methods finding the same source-driven partition shows that the dominant structure in the VGG embedding is technical. This is a property of the data, not a failure of the algorithms.

### Overall take-away
1. **PCA** identifies the batch axes
2. **Projections:** t-SNE and UMAP produce visibly clean islands, but they are source islands: the neighbour agreement with source is 0.95-0.96 versus 0.69-0.70 with label (PCA 0.67 vs 0.50), and the silhouette of label is negative in all three. 
Visual separation must be validated numerically and against metadata.
1. **Supervised models** reach 0.79 balanced accuracy within the pooled data, but only chance level on a held-out source.
2. **Clustering** recovers the sources, not the classes.

Any biological claim should be tested with source held out, and the analysis would benefit from batch correction (e.g. per-source centring/standardisation, or including source as a covariate) and from balanced sampling of classes across sources before training.
